# **E-Commerce Sales Analysis**

```python
Project Objective

The main objective of this project is to analyze e-commerce sales data using Python, Pandas, Matplotlib, and Seaborn to identify sales trends, product performance, customer purchasing behavior, and profitability. The project aims to transform raw sales data into meaningful insights that help businesses understand their performance and make data-driven decisions.

Specific objectives
1. Data cleaning: Identify missing values, duplicate records, invalid dates, and incorrect data.
2. Sales analysis: Calculate total sales and analyze monthly sales trends.
3. Profitability analysis: Evaluate total profit and compare profit across product categories.
4. Product performance: Identify top-selling products and products generating the highest sales.
5. Customer analysis: Identify top customers based on their purchase value.
6. Category analysis: Compare sales and profitability across different categories.
7. Business insights: Analyze payment methods, order statuses, and regional sales performance.
8. Data visualization: Create charts to present findings clearly and support business decision-making.

Expected outcome: A Python-based sales analysis project that converts raw e-commerce data into useful business insights, with visualizations and exported summary reports.
```

```python
# Top 10 Products by Sales
top_products = (
    df.groupby("Product_Name")["Total_Sales"]
      .sum()
      .sort_values(ascending=False)
      .head(10)
)

print(top_products)
```

```python
# Top 10 Products by Quantity
top_quantity = (
    df.groupby("Product_Name")["Quantity"]
      .sum()
      .sort_values(ascending=False)
      .head(10)
)

print(top_quantity)
```

```python
# CATEGORY ANALYSIS
```

```python
# Category-wise Sales
category_sales = (
    df.groupby("Category")["Total_Sales"]
      .sum()
      .sort_values(ascending=False)
)

print(category_sales)
```

```python
# Category-wise Profit
category_profit = (
    df.groupby("Category")["Profit"]
      .sum()
      .sort_values(ascending=False)
)

print(category_profit)
```

```python
# MONTHLY SALES TREND
# Keep only valid dates for monthly analysis
monthly_df = df[df["Order_Date"].notna()].copy()
monthly_df
```

```python
# Group sales by month number
monthly_sales = monthly_df.groupby(monthly_df["Order_Date"].dt.month)["Total_Sales"].sum()
monthly_sales
```

```python
monthly_df["Month"] = monthly_df["Order_Date"].dt.month_name()
```

```python
monthly_profit = monthly_df.groupby("Month")["Profit"].sum()

print(monthly_profit)
```

```python
# Monthly Orders
monthly_orders = (monthly_df.groupby("Month")["Order_ID"].count())

print(monthly_orders)
```

```python
# Calculate monthly sales and profit
monthly_sales = monthly_df.groupby("Month").agg(
    Total_Sales=("Total_Sales", "sum"),
    Total_Profit=("Profit", "sum")
).reset_index()

monthly_sales
```

```python
month_order = [
    "January", "February", "March", "April",
    "May", "June", "July", "August",
    "September", "October", "November", "December"
]

monthly_profit = monthly_profit.reindex(month_order)

print(monthly_profit)
```

```python
plt.figure(figsize=(10, 5))
monthly_profit.plot(marker="o")

plt.title("Monthly Profit Trend")
plt.xlabel("Month")
plt.ylabel("Total Profit")
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```
